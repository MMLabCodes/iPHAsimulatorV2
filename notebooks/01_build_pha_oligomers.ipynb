# 01 Build PHA Oligomers

This notebook introduces the basic building step in iPHASimulator. It is written for users who may not be computational chemists or Python developers.

A **PHA oligomer** is a short model of a longer PHA polymer chain. In this project, `degree` means the number of repeat units. For example, `degree = 4` means a four-repeat-unit oligomer.

The important idea is simple: the notebook asks the package code in `src/iphasimulator` to build molecules. The notebook itself should only guide the workflow.

## Step 1: Load the builder tools

The curated validation set uses three representative PHAs:

- **PHB**: short methyl side chain
- **PHO**: medium alkyl side chain
- **PHDD**: longer alkyl side chain

These are useful first checks because they cover small, medium, and larger molecules.

```python
from iphasimulator.workflows import DEFAULT_VALIDATION_TARGETS, build_validation_molecules

DEFAULT_VALIDATION_TARGETS
```

## Step 2: Build the default molecules

The next cell creates RDKit molecule objects. You do not need to understand RDKit to use this notebook; RDKit is the chemistry engine used behind the scenes.

```python
molecules = build_validation_molecules(DEFAULT_VALIDATION_TARGETS)
list(molecules)
```

## Step 3: Optional larger systems

Larger oligomers are useful after the small systems work. They are not the default because later AmberTools charge generation can become slow for larger molecules.

```python
from iphasimulator.workflows import LARGE_VALIDATION_TARGETS

large_molecules = build_validation_molecules(LARGE_VALIDATION_TARGETS)
list(large_molecules)
```
