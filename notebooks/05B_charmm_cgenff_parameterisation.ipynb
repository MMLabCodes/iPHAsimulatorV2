# 05B CHARMM/CGenFF Parameterisation

This notebook is a documentation placeholder for a future CHARMM/CGenFF parameterisation workflow. It does not automate CGenFF or write simulation-ready files yet.

Use this notebook when you want to prepare a polymer or small-molecule ligand for the CHARMM force-field ecosystem, especially when the next step is CHARMM-GUI Solution Builder.

## Intended CHARMM-GUI route

The manual route is:

1. Export a chemically correct input structure from notebook `04_export_structures.ipynb`.
2. Upload the structure to CHARMM-GUI Ligand Reader & Modeler.
3. Generate or inspect CGenFF parameters for the ligand or polymer fragment.
4. Download the generated CHARMM-compatible files.
5. Use those files later in CHARMM-GUI Solution Builder when constructing a solvated simulation system.

This route is separate from notebook `05A_amber_gaff2_parameterisation.ipynb`, which keeps the existing AmberTools/GAFF2 workflow unchanged.

## CHARMM-GUI Ligand Reader & Modeler

CHARMM-GUI Ligand Reader & Modeler prepares non-standard molecules for CHARMM-family simulations. For molecules that are not already available in the CHARMM force field, it can use CGenFF-style atom typing and parameter assignment.

Use it to check whether the uploaded structure is interpreted correctly before moving to a full solvated system. Pay particular attention to atom names, residue names, total charge, protonation state, stereochemistry, and any warnings about missing or uncertain parameters.

## Expected input structures

Common accepted ligand inputs include:

- `.sdf`: preferred when bond orders and stereochemistry are encoded clearly.
- `.mol2`: useful when atom names, residue names, charges, and bond orders are already curated.
- `.pdb`: useful for coordinates, but often weaker for bond-order information unless paired with additional chemistry-aware input.

For this project, start from the structures exported by notebook `04_export_structures.ipynb`. If several formats are available, prefer the one that best preserves the intended chemical graph rather than the one that merely looks correct in a viewer.

## Stereochemistry and bond orders

CGenFF parameter assignment depends on the molecular graph. A structure with correct 3D coordinates can still be unsuitable if the file loses bond orders, aromaticity, formal charge, or stereochemical labels.

Before submitting to CHARMM-GUI, check:

- Chiral centers match the intended PHA stereochemistry.
- Ester and carbonyl bonds have the correct single/double-bond pattern.
- Terminal groups and protonation states match the intended simulation system.
- The total molecular charge is the charge you expect.
- No atoms are missing hydrogens required for the selected protonation state.

If CHARMM-GUI reports unexpected chemistry, fix the input structure before using the generated parameters downstream.

## CGenFF parameter generation

CGenFF assigns atom types, bonded parameters, and non-bonded parameters for small organic molecules compatible with CHARMM simulations. The output may include penalty scores or warnings that identify parameters inferred with lower confidence.

For a production workflow, review high-penalty terms rather than treating the generated files as automatically final. This placeholder notebook records the intended manual workflow only; it does not run CGenFF locally or through a web service.

## Generated files to keep

A CHARMM-GUI/CGenFF preparation may produce files such as:

- `.str`: CHARMM stream file containing topology and parameter information for the ligand.
- `.psf`: CHARMM protein structure file describing atoms, bonds, angles, dihedrals, and connectivity.
- `.pdb`: coordinate file with CHARMM-compatible atom and residue naming.

Keep these files together with the exact input structure used to generate them. The `.str`, `.psf`, and `.pdb` files should be treated as a matched set unless CHARMM-GUI documentation for a later step says otherwise.

## Later use in CHARMM-GUI Solution Builder

After Ligand Reader & Modeler has produced suitable ligand files, those files can be supplied to CHARMM-GUI Solution Builder to construct a solvated system with water, ions, periodic box settings, and engine-specific inputs.

The expected future workflow is:

1. Prepare and validate ligand parameters with Ligand Reader & Modeler.
2. Upload or select the generated ligand files in Solution Builder.
3. Build the solvated CHARMM-compatible system.
4. Download the generated simulation input package.
5. Run or adapt the generated inputs outside this placeholder notebook.

Automation for these steps can be added later without changing the existing Amber/GAFF2 workflow.
